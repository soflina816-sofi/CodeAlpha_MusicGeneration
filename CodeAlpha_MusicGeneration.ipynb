```python
from music21 import corpus, note, chord, stream, midi
import numpy as np

# 1. Load a sample MIDI file from the music21 built-in corpus (e.g., Bach chorale)
print("Loading sample MIDI data from music21 corpus...")
bach_bach = corpus.parse('bach/bwv66.6')

# 2. Extract notes and chords
notes = []
for element in bach_bach.flat.notes:
    if isinstance(element, note.Note):
        notes.append(str(element.pitch))
    elif isinstance(element, chord.Chord):
        notes.append('.'.join(str(n) for n in element.normalOrder))

print(f"Total notes extracted: {len(notes)}")
print("Sample extracted notes:", notes[:10])

# 3. Create a simple generated melody sequence (Simulating AI generation output for demonstration)
# In a full deep learning model, you would feed these into an LSTM/RNN or GAN[span_3](start_span)[span_3](end_span).
generated_notes = notes[:32] # Taking a sequence to convert back to MIDI

# 4. Convert generated sequences back to a MIDI file[span_4](start_span)[span_4](end_span)
output_stream = stream.Stream()
for pattern in generated_notes:
    if '.' in pattern:
        # If it's a chord
        chord_notes = [int(n) for n in pattern.split('.')]
        new_chord = chord.Chord(chord_notes)
        output_stream.append(new_chord)
    else:
        # If it's a single note
        new_note = note.Note(pattern)
        output_stream.append(new_note)

# Save the generated music to a MIDI file
output_file = "generated_music.mid"
output_stream.write('midi', fp=output_file)
print(f"Generated music successfully saved as '{output_file}'!")
```
